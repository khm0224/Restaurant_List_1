```python
# 교동 음식점 지오코딩: ArcGIS 방식
import time
import pandas as pd
import requests

INPUT_FILE = "filter_file_Gyo_dong.csv"
OUTPUT_FILE = "filter_file_Gyo_dong_geocoded.csv"

df = pd.read_csv(INPUT_FILE, encoding="utf-8-sig")

# 좌표 열은 반드시 숫자형으로 생성
df["위도"] = pd.Series(float("nan"), index=df.index, dtype="float64")
df["경도"] = pd.Series(float("nan"), index=df.index, dtype="float64")
df["변환주소"] = pd.Series(pd.NA, index=df.index, dtype="string")
df["지오코딩상태"] = pd.Series(pd.NA, index=df.index, dtype="string")
df["매칭점수"] = pd.Series(float("nan"), index=df.index, dtype="float64")

API_URL = (
    "https://geocode.arcgis.com/arcgis/rest/services/"
    "World/GeocodeServer/findAddressCandidates"
)

session = requests.Session()
session.headers["User-Agent"] = "Mozilla/5.0 restaurant-list-geocoder/1.0"

# 같은 주소를 중복 조회하지 않기 위한 캐시
cache = {}

for number, index in enumerate(df.index, start=1):
    address = " ".join(str(df.at[index, "지번주소"]).strip().split())

    if not address or address.lower() == "nan":
        result = {"상태": "주소없음"}

    elif address in cache:
        result = cache[address]

    else:
        try:
            response = session.get(
                API_URL,
                params={
                    "SingleLine": address,
                    "f": "json",
                    "maxLocations": 1,
                    "outFields": "Match_addr",
                    "countryCode": "KOR",
                    "forStorage": "false",
                    "outSR": 4326,
                },
                timeout=30,
            )
            response.raise_for_status()
            data = response.json()

            if data.get("error"):
                message = data["error"].get("message", "ArcGIS API 오류")
                result = {"상태": f"API오류: {message}"}

            else:
                candidates = data.get("candidates", [])

                if candidates:
                    candidate = candidates[0]
                    location = candidate["location"]
                    score = float(candidate.get("score", 0))

                    result = {
                        "상태": "성공",
                        "위도": float(location["y"]),
                        "경도": float(location["x"]),
                        "변환주소": candidate.get("address", ""),
                        "매칭점수": score,
                    }
                else:
                    result = {"상태": "검색결과없음"}

        except requests.RequestException as error:
            result = {"상태": f"요청오류: {error}"}
        except (KeyError, TypeError, ValueError) as error:
            result = {"상태": f"응답처리오류: {error}"}

        cache[address] = result
        time.sleep(0.25)

    df.at[index, "지오코딩상태"] = result["상태"]

    if result["상태"] == "성공":
        df.at[index, "위도"] = result["위도"]
        df.at[index, "경도"] = result["경도"]
        df.at[index, "변환주소"] = result["변환주소"]
        df.at[index, "매칭점수"] = result["매칭점수"]

    print(
        f"[{number}/{len(df)}] "
        f"{address} -> {result['상태']}"
    )

    # 실행 중단에 대비한 중간 저장
    if number % 10 == 0:
        df.to_csv(
            OUTPUT_FILE,
            index=False,
            encoding="utf-8-sig",
        )

df.to_csv(
    OUTPUT_FILE,
    index=False,
    encoding="utf-8-sig",
)

print("\n지오코딩 완료")
print(df["지오코딩상태"].value_counts(dropna=False))
print("결과 파일:", OUTPUT_FILE)

display(
    df[
        [
            "사업장명",
            "지번주소",
            "위도",
            "경도",
            "변환주소",
            "매칭점수",
            "지오코딩상태",
        ]
    ].head(10)
)
```
