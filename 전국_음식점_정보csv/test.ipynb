```python
import pandas as pd

# 1. CSV 파일 불러오기
# (공공데이터는 보통 'cp949' 또는 'euc-kr' 인코딩을 사용합니다. 에러 발생 시 'utf-8'로 변경하세요.)
file_path = 'gangwon_filtered.csv'
df = pd.read_csv(file_path)

# 2. '도로명주소' 열에서 '강원'이 포함된 행 필터링 (결측치 예방 na=False)
# filtered_df = df[df['도로명주소'].str.contains('강원특별자치도 춘천시', na=False)]
filtered_df = df[df['영업상태명'].str.contains('영업/정상', na=False)]

# 3. 필터링된 결과 확인
print(f"전체 데이터 수: {len(df)}건")
print(f"강원 지역 데이터 수: {len(filtered_df)}건")

# 4. 필터링 결과 저장 (엑셀에서 바로 열 때 깨지지 않도록 utf-8-sig 사용)
filtered_df.to_csv('gangwon_filtered.csv', index=False, encoding='utf-8-sig')
```

```python
uptae_list = filtered_df['업태구분명'].dropna().drop_duplicates().tolist()
print(uptae_list)
uptae_list.to_csv('title.csv', index=False, encoding='utf-8-sig')
```

```python
gita_df = filtered_df[filtered_df['업태구분명'] == "기타"]
print(f"'기타' 업태 데이터 수: {len(gita_df)}건")
print(gita_df)
```

```python
result_df = gita_df[['업태구분명', '사업장명', '지번주소']]
print(result_df)
result_df.to_csv('other.csv', index=False, encoding='utf-8-sig')
```

```python
import pandas as pd

sssss = 'gangwon_filtered.csv'
df = pd.read_csv(sssss)

exclude_list = [
    "정종/대포집/소주방",
    "푸드트럭",
    "감성주점",
    "패스트푸드",
    "키즈카페",
    "패밀리레스트랑",
    "외국음식전문점(인도,태국등)"
]

result_df = df[['업태구분명', '사업장명', '지번주소']]
filtered_df = result_df[~result_df['업태구분명'].isin(exclude_list)]

print(filtered_df.head())
filtered_df.to_csv('filter_file.csv', index=False, encoding='utf-8-sig')
```
